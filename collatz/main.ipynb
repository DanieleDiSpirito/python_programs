```python
import matplotlib.pyplot as plt

def collatz(number):
    sequence = []
    while number != 1:
        sequence.append(number)
        if number % 2 == 0:
            number = number // 2
        else:
            number = 3 * number + 1
    sequence.append(1)
    return sequence


plt.figure(figsize=(15, 30))
counter = 0
Max = max(collatz(27)) # 9232 appears first at 27
for i in range(1, 100):
    if max(collatz(i)) >= Max:
        # plot on different graphs but same window
        plt.subplot(16, 1, counter + 1)
        plt.plot(collatz(i), label=f'{i}')
        plt.xlim(0)
        plt.ylim(0)
        plt.legend(loc='upper right')
        counter += 1
        Max = max(collatz(i))

plt.show()

print(f"{Max=}")
```

```python
collatz_values = []
for i in range(1, 100):
    collatz_values.extend(collatz(i))

plt.hist(collatz_values, bins=50, color="blue", edgecolor="black")
```
